Writing simple business report using python

```python

import pandas as pd

sales = pd.DataFrame({
    "Product": ["Shoes", "Bags", "Dresses"],
    "Revenue": [9000, 7500, 12500],
    "Cost": [5400, 3500, 7000]
})

sales["Gross_Profit"] = sales["Revenue"] - sales["Cost"]
sales["Gross_Margin_Percent"] = (
    sales["Gross_Profit"] / sales["Revenue"] * 100
)

total_revenue = sales["Revenue"].sum()
total_cost = sales["Cost"].sum()
total_profit = sales["Gross_Profit"].sum()
overall_margin = total_profit / total_revenue * 100

highest_revenue = sales.loc[sales["Revenue"].idxmax()]
highest_margin = sales.loc[
    sales["Gross_Margin_Percent"].idxmax()
]

print("BOUTIQUE SALES REPORT")
print("---------------------")
print(f"Total revenue: KSh {total_revenue:,.0f}")
print(f"Total product costs: KSh {total_cost:,.0f}")
print(f"Total gross profit: KSh {total_profit:,.0f}")
print(f"Overall gross margin: {overall_margin:.2f}%")

print("\nKEY FINDINGS")
print(f"Highest revenue: {highest_revenue['Product']}")
print(f"Highest gross margin: {highest_margin['Product']}")

print("\nRECOMMENDATION")
print(
    "Investigate demand and stock levels for bags, "
    "which have the highest gross margin."
)
```

Creating summary tables

```python

import pandas as pd

sales = pd.DataFrame({
    "Product": ["Shoes", "Bags", "Dresses"],
    "Revenue_KSh": [9000, 7500, 12500],
    "Cost_KSh": [5400, 3500, 7000]
})

sales["Gross_Profit_KSh"] = (
    sales["Revenue_KSh"] - sales["Cost_KSh"]
)

sales["Gross_Margin_Percent"] = (
    sales["Gross_Profit_KSh"] / sales["Revenue_KSh"] * 100
)

report = sales.sort_values(
    "Gross_Margin_Percent",
    ascending=False
).round(2)

print("PRODUCT PERFORMANCE REPORT")
print("=" * 65)
print(report.to_string(index=False))
```
